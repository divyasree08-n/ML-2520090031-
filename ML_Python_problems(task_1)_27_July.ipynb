```python
n = int(input("Enter a number:"))
reverse = 0

while n > 0:
  digit = n % 10
  reverse = reverse * 10 + digit
  n = n//10
print("Reversed Number=", reverse)
```

```python
num = 6

if num <= 1:
    print(num, "is not a prime number")
else:
    is_prime = True
    i = 2

    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
```
